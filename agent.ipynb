```python
from langchain.tools import tool
from firecrawl import Firecrawl
from typing import Dict,Any

firecrawl = Firecrawl()

@tool
def top_course (query:str)-> str:
    "Return the top 10 websites for each course search "

    result =  firecrawl.search (query)
    return result.web[:10]



```

```python
r = top_course.invoke("python")
print(r)
```
